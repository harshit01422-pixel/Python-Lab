```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
data = {
    'Days':['Mon','Tue','Wed','Thu','Fri'],
    'Study_Hours':[2,3,2.5,4,3.5],
    'Marks':[50,60,55,70,65],
    'Attendance':[60,70,65,80,75],
    'Sleep_Hours':[6,7,6,8,7],
    'Assignments_Completed':[3,4,3,5,6]
}
df = pd.DataFrame(data)
df
```

```python
#1 Line chart (trend)
plt.plot(df['Days'],df['Study_Hours'],marker='o',linestyle='-.',color = 'green')
plt.title('Study Hours Over Days')
plt.xlabel('Days')
plt.ylabel('Study Hours')
plt.show()
```

```python
# Advnaced Line chart
plt.figure(figsize=(7,4))
plt.plot(df['Days'],df['Study_Hours'],marker = 'd', linestyle = '-.', color = 'green', label = 'Study Hours')
plt.plot(df['Days'],df['Marks'], marker = '*', linestyle = '--',color = 'red',label = 'Marks')
plt.title("Study Hours and Marks Over Days")
plt.xlabel('Days')
plt.ylabel('Values')
plt.legend()
plt.show()
```

```python
#2 Bar chart (Comparison)
plt.bar(df['Days'],df['Marks'],color = '#900603') #Ruby Red
plt.title('Marks Comparison')
plt.xlabel('Days')
plt.ylabel('Marks')
plt.show()
```

```python
plt.figure(figsize=(7,4))
plt.bar(df['Study_Hours'],df['Marks'],color = '#f0dd31',edgecolor = 'black',width=0.3)
plt.title('Study Hours vs Marks')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.show()
```

```python
# Advanced Bar Graph
plt.figure(figsize=(7,4))
bars = plt.bar(df['Days'],df['Marks'],color = 'cyan')
for bar in bars:
  y = bar.get_height()
  plt.text(bar.get_x() + bar.get_width()/2,y,str(y), ha = 'center',va = 'bottom')
plt.title('Marks Comparison')
plt.xlabel('Days')
plt.ylabel('Marks')
plt.grid(axis = 'x')
plt.show()
```

```python
# Comparison
plt.figure(figsize=(7,4))
import numpy as np
x = np.arange(len(df['Days']))
width = 0.35
plt.bar(x-width/2, df['Study_Hours'],width,label = 'Study Hours',color = 'green')
plt.bar(x+width/2,df['Marks'],width,label = 'Marks',color = 'red')
plt.xticks(x,df['Days'])
plt.title("Study hours vs Marks")
plt.xlabel("Days")
plt.ylabel("Values")
plt.legend()
plt.show()
```

```python
#3.Histogram (Distribution)
#Shows how values are spread
plt.figure(figsize=(7,4))
import numpy as np

data = {
    'Marks':[50,60,55,70,65,75,80,85,90,45,68,72,77,82,88]
}
df = pd.DataFrame(data)
plt.hist(df['Marks'],bins=3,color='cyan',edgecolor='black',alpha=1) #Alpha = opacity, range = 0 to 1
plt.title('Marks Distribution')
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(7,4))
data = {
    'Days':['Mon','Tue','Wed','Thu','Fri'],
    'Study_Hours':[2,3,2.5,4,3.5],
    'Marks':[50,60,55,70,65],
    'Attendance':[60,70,65,80,75],
    'Sleep_Hours':[6,7,6,8,7],
    'Assignments_Completed':[3,4,3,5,6]
}
df = pd.DataFrame(data)
plt.scatter(df['Study_Hours'],df['Marks'],color = '#900603')
plt.title("Study Hours vs Marks")
plt.xlabel("Study hours")
plt.ylabel("Marks")
plt.grid()
plt.show()
```

```python
# Conditional Scatter plot
# Example Category
plt.figure(figsize=(7,4))
df['Result'] = ['Fail','Pass','Fail','Pass','Pass']
colors = ['red' if r=='Fail' else 'green' for r in df['Result']]
plt.scatter(df['Study_Hours'], df['Marks'], c = colors)
plt.title("Study hours vs Marks (Result-wise)")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

```python
# Dataset-2
data = {
    'Day': ['Mon','Tue','Wed','Thu','Fri','Sat','Sun','Mon','Tue','Wed'],
    'Region': ['North','South','East','West','North','South','East','West','North',
               'South'],
    'Sales': [200,220,250,270,300,350,400,280,260,290],
    'Profit': [20,25,30,35,50,60,80,40,38,45],
    'Customers': [30,35,40,45,50,65,80,55,48,52],
    'Category': ['Food','Clothing','Electronics','Food','Clothing','Electronics',
                 'Food','Clothing','Electronics','Food']
}

df = pd.DataFrame(data)
print(df)
```

# Same graph using Seaborn (For better Visualization)

```python
# Line Plot
plt.figure(figsize = (7,4))
sns.lineplot(x='Day',y='Sales',data =df)
plt.title("Seaborn Line Plot")
plt.show()
```

```python
#Bar Plot
sns.barplot(x='Day', y='Profit', data=df)
plt.title("Seaborn Bar Plot")
plt.show()
```

```python
#Histogram
sns.histplot(df['Sales'], bins=5)
plt.title("Seaborn Histogram")
plt.show()
```

```python
#Scatter Plot
sns.scatterplot(x='Sales', y='Profit', data=df)
plt.title("Seaborn Scatter Plot")
plt.show()
```
