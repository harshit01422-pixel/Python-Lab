```python
import pandas as pd
import numpy as np
data = pd.read_csv("/content/covid_19_data.csv")
data
```

```python
data['ObservationDate'] = data['ObservationDate'].astype('datetime64[ns]')
data['Confirmed'] = data['Confirmed'].astype('int64')
data['Deaths'] = data['Deaths'].astype('int64')
data['Recovered'] = data['Recovered'].astype('int64')
data
```

```python
# Active Cases
data['Active'] = data['Confirmed'] - data['Recovered'] - data['Deaths']
data.head(50)
data.iloc[50:100] #gives values from 50 to 99
```

```python
import json
import pandas as pd
import plotly.express as px

# Load GeoJSON
geojson_path = '/content/india (1).json'

with open(geojson_path, 'r') as f:
    india_states_geojson = json.load(f)

print("GeoJSON loaded successfully!")

# Inspect property keys (IMPORTANT)
print("\nAvailable keys in GeoJSON properties:")
print(india_states_geojson['features'][0]['properties'].keys())


# ---------------------------
# FIX STATE NAME MISMATCHES
# ---------------------------
# Based on repeated output, the previous `replace` operations were converting
# state names to versions that were *not* present in the GeoJSON's 'st_nm' property.
# This means the GeoJSON likely uses the original names from the DataFrame, or slightly different ones.
# Let's remove the problematic replacements and only keep the 'Unknown' cleanup for now.
# The 'States NOT matching GeoJSON' will then show the *actual* mismatches from the DataFrame's original names.

top_state['Province/State'] = top_state['Province/State'].replace({
    'Unknown': 'None' # Clean up 'Unknown' entries, which won't map anyway.
})


# ---------------------------
# DEBUG: CHECK MISSING STATES
# ---------------------------
geo_states = [f['properties']['st_nm'] for f in india_states_geojson['features']]
data_states = top_state['Province/State'].unique()

missing = set(data_states) - set(geo_states)

print("\nStates NOT matching GeoJSON:")
print(missing)


# ---------------------------
# PLOT MAP
# ---------------------------
fig = px.choropleth(
    top_state,
    geojson=india_states_geojson,
    locations='Province/State',
    featureidkey='properties.st_nm',
    color='Confirmed',
    hover_name='Province/State',
    color_continuous_scale='Viridis',
    title='COVID-19 Confirmed Cases in Indian States'
)

# Zoom to India and hide axes
fig.update_geos(fitbounds="locations", visible=False)

# Optional: show borders for debugging
fig.update_traces(marker_line_width=0.5, marker_line_color='black')

fig.show()
```
